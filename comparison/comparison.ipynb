```python
import pandas as pd
import seaborn as sns
from pathlib import Path
import matplotlib.pyplot as plt
```

```python
path = Path.cwd().parent / 'results/comparision_table.csv'
save_path = Path.cwd().parent / 'results/gbn_vs_sr_graph.png'
path.parent.mkdir(parents=True, exist_ok=True)
```

```python
df = pd.read_csv(path)
```

```python
df
```

```python
plt.figure(figsize=(10,6))

ax = sns.barplot(data=df, x='Scenario', y='Transmitted', hue='Protocol')

plt.title('GBN vs. SR: Total Network Load Across Error Scenarios', fontsize=14, fontweight='bold')
plt.xlabel('Network Condition', fontsize=12, fontweight='bold')
plt.ylabel('Total Packets Transmitted', fontsize=12, fontweight='bold')
plt.legend(title='Protocol', title_fontsize=11)

for container in ax.containers:
    ax.bar_label(container, padding=3, fontweight='bold')


plt.savefig(save_path, dpi=300)
print(f'Graph successfully saved to {save_path}')
```
